```python
### Example: load an EDF file

import mne
from pathlib import Path

def load_edf(edf_path: str | Path,
             preload: bool = True,
             verbose: str = "ERROR") -> mne.io.Raw:
    """
    Load a single EDF file into an MNE Raw object.
    """
    edf_path = Path(edf_path)
    raw = mne.io.read_raw_edf(edf_path, preload=preload, verbose=verbose)
    return raw

# Usage example
edf_path = "sub-S0001123127549_ses-1_task-EEG_eeg.edf"
raw = load_edf(edf_path)
```
